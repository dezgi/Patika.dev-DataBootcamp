```python
import pandas as pd

df_2016 = pd.read_csv('2016.csv')

# Hangi ülkeler veya bölgeler genel mutlulukta ve mutluluğa katkıda bulunan altı faktörün her birinde en yüksek sırada yer alıyor?

# most happiest country
most_happy_country = df_2016.loc[df_2016['Happiness Score'].idxmax(), 'Country']
print(f"En mutlu ülke (2016): {most_happy_country}")

# having best score based on these 6 labels
factors = {
    'Economy (GDP per Capita)': 'Ekonomi',
    'Family': 'Sosyal Destek',
    'Health (Life Expectancy)': 'Yaşam Süresi',
    'Freedom': 'Özgürlük',
    'Trust (Government Corruption)': 'Yolsuzluk Algısı (Ters)',
    'Generosity': 'Cömertlik'
}

print("\nFaktör Bazlı En Yüksek Skora Sahip Ülkeler:\n")
for col, label in factors.items():
    country = df_2016.loc[df_2016[col].idxmax(), 'Country']
    score = df_2016[col].max()
    print(f"{label}: {country} ({score:.3f})")
```

```python
# 2015-2016 ile 2016-2017 raporları arasında ülke sıralamaları veya puanları nasıl değişti?

df_2015 = pd.read_csv('2015.csv')
df_2016 = pd.read_csv('2016.csv')

ranking_change = pd.merge( # combine two datas
    df_2015[['Country', 'Happiness Rank']],
    df_2016[['Country', 'Happiness Rank']],
    on='Country',
    suffixes=('_2015', '_2016')
)


ranking_change['Rank_Change'] = ranking_change['Happiness Rank_2015'] - ranking_change['Happiness Rank_2016'] # calculate rank change


print("\n Sıralamada En Çok Yükselen 10 Ülke:\n")
print(ranking_change.sort_values('Rank_Change', ascending=False).head(10))

print("\n Sıralamada En Çok Düşen 10 Ülke:\n")
print(ranking_change.sort_values('Rank_Change').head(10))
```

```python
# Herhangi bir ülke mutlulukta önemli bir artış veya azalma yaşadı mı?

score_change = pd.merge( # combine two datas
    df_2015[['Country', 'Happiness Score']],
    df_2016[['Country', 'Happiness Score']],
    on='Country',
    suffixes=('_2015', '_2016')
)


score_change['Score_Change'] = score_change['Happiness Score_2016'] - score_change['Happiness Score_2015']  # calculate score change

print("\n Mutluluk Skorunda En Fazla Artış Gösteren 10 Ülke:\n")
print(score_change.sort_values('Score_Change', ascending=False).head(10))

print("\n Mutluluk Skorunda En Fazla Azalma Gösteren 10 Ülke:\n")
print(score_change.sort_values('Score_Change').head(10))
```
